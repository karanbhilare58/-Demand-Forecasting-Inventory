```python
import pandas as pd
import numpy as np

# Load original dataset
df = pd.read_csv("../data/raw/train.csv")

print("Dataset loaded successfully")
print("Shape:", df.shape)
```

```python
print(df.dtypes)
```

```python
df["date"] = pd.to_datetime(df["date"])

print(df.dtypes)
```

```python
df = df.sort_values(
    by=["store", "item", "date"]
).reset_index(drop=True)

df.head()
```

```python
print(df.isnull().sum())
```

```python
duplicates = df.duplicated(
    subset=["date", "store", "item"]
).sum()

print("Duplicate store-item-date records:", duplicates)
```

```python
print(df["sales"].describe())

print("Minimum sales:", df["sales"].min())
print("Maximum sales:", df["sales"].max())

print("Zero-sales records:", (df["sales"] == 0).sum())
print("Negative-sales records:", (df["sales"] < 0).sum())
```

```python
expected_days = df["date"].nunique()

coverage = df.groupby(
    ["store", "item"]
).agg(
    total_records=("date", "count"),
    unique_dates=("date", "nunique"),
    first_date=("date", "min"),
    last_date=("date", "max")
)

print("Expected dates per combination:", expected_days)
print("Minimum records:", coverage["total_records"].min())
print("Maximum records:", coverage["total_records"].max())

print("Incomplete combinations:")
print(coverage[
    coverage["total_records"] != expected_days
])
```

```python
# Save processed dataset
df.to_csv(
    "../data/processed/sales_cleaned.csv",
    index=False
)

print("Cleaned dataset saved successfully")
```
